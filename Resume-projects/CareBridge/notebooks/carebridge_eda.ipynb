# CareBridge exploratory data analysis
This notebook demonstrates the NumPy → Pandas → Matplotlib → SQL part of the project using entirely synthetic data.

```python
import sqlite3
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
connection = sqlite3.connect(ROOT / 'data' / 'carebridge.db')
```

```python
tasks = pd.read_sql_query('SELECT * FROM preparation_tasks', connection)
symptoms = pd.read_sql_query('SELECT * FROM symptoms', connection)
display(tasks.head(), symptoms.describe(include='all'))
```

```python
applicable = tasks.query("status != 'not_applicable'")
score = np.rint(np.mean(applicable.status.to_numpy() == 'complete') * 100).astype(int)
print(f'Appointment Preparation Score: {score}% (administrative, not medical)')
```

```python
ax = symptoms.sort_values('onset_date').plot.bar(x='symptom', y='severity', color='#277b73', legend=False)
ax.set_ylim(0, 10); ax.set_ylabel('Patient-rated severity'); ax.set_title('Synthetic symptom overview')
plt.show()
```
